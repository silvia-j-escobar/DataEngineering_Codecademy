```python
# Silvia
# Len's Slice

# 1. Create a list called toppings
toppings = ["pepperoni", "pineapple", "cheese", "sausage", "olives", "anchovies", "mushrooms"]

# 2. Create a list called prices to keep track of how much each kind of pizza slice costs
prices = [2, 6, 1, 3, 2, 7, 2]

# 3. Count the number of 2 in the prices list, and store the results in num_two_dollar_slices. Print out.
num_two_dollar_slices = prices.count(2)
print(num_two_dollar_slices)

# 4. Find the length of toppings and store it in a variable called num_pizzas
num_pizzas = len(toppings)

# 5. Print the string We sell [num_pizzas] different kinds of pizza!, where [num_pizzas] represents the value of our variable num_pizzas.

print("We sell " + str(num_pizzas) + " different kinds of pizza!")

# 6. Create a new two-dimensional list called pizza_and_prices
pizza_and_prices = [[2, "pepperoni"], [6, "pineapple"], [1, "cheese"], [3, "sausage"], [2, "olives"], [7, "anchovies"], [2, "mushrooms"]]

# 7. Print pizza_and_prices
print(pizza_and_prices)

#8. Sort pizza_and_prices
pizza_and_prices.sort()

# 9. Store the first element of pizza_and_prices in a variable calles cheapest_pizza
cheapest_pizza = pizza_and_prices[0]

# 10. Get the last item of the pizza_and_prices list and store it in a variable calles priciest_pizza
priciest_pizza = pizza_and_prices[-1]

# 11. Remove "anchovies" from pizza_and_prices list
pizza_and_prices.remove([7, "anchovies"])

# 12. Add "peppers" to toppings list
pizza_and_prices.insert(4, [2.5, "peppers"])

# 13. Slice the pizza_and_prices list and store the three lowest-cost pizzas in a list called three_cheapest.
three_cheapest = pizza_and_prices[:3]

# 14. Print the three_cheapest list.
print(three_cheapest)
```
