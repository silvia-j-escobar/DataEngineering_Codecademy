You are the Data Analyst at Carly’s Clippers, the newest hair salon on the block. Go through the lists of data that have been collected over the past couple of weeks. You will be calculating some important metrics that Carly can use to plan out the operation of the business for the rest of the month.

Three lists have been provided:

hairstyles: the names of the cuts offered at Carly’s Clippers
prices: the price of each hairstyle in the hairstyles list
last_week: the number of purchases for each hairstyle type in the last week
Each index in hairstyles corresponds to the associated index in prices and last_week.

For example, the hairstyle "bouffant" has an associated price of 30 from the prices list, and was purchased 2 times in the last week, as shown in the last_week list. Each of these elements is at the first index of its respective list.

```python
# Lists provided
hairstyles = ["bouffant", "pixie", "dreadlocks", "crew", "bowl", "bob", "mohawk", "flattop"]

prices = [30, 25, 40, 20, 20, 35, 50, 35]

last_week = [2, 3, 5, 8, 4, 4, 6, 2]
```

```python
# Carly wants to market her low prices. We want to find out what the average price of a haircut is.
# First, let’s sum up all the prices of the haircuts. Create a variable total_price, and set it to 0.
# Create variable total_price
total_price = 0
```

```python
# 2. Loop trhough prices list and add each price to the variable total_price
for price in prices:
  total_price += price
```

```python
# 3. Create a variable called average_price that is equal to the total_price divided by the number of prices. You can get the number of prices by using the len() function.
average_price = total_price / len(prices)
```

```python
# 4. Print the value of average_price
print(average_price)
```

```python
# 5. That average price is more expensive than Carly thought it would be! She wants to reduce all prices by five dollars.
# Use a list comprehension to create a list called new_prices, which has each element in prices minus 5.
new_prices = [price - 5 for price in prices]
```

```python
# 6. Print new_prices
print(new_prices)
```

Revenue:

```python
# 7. Carly really wants to make sure that Carly’s Clippers is a profitable endeavor. She first wants to know how much revenue was brought in last week.
# Create a variable called total_revenue and set it to 0.
total_revenue = 0
```

```python
# 8. Use a for loop to create a variable i that goes from 0 to len(hairstyles).
# 9. Add the product of prices[i] (the price of the haircut at position i) and last_week[i] (the number of people who got the haircut at position i)
# to total_revenue at each step.
# 10. Print total_revenue

for i in range(len(hairstyles)):
    total_revenue += prices[i] * last_week[i]

print(total_revenue)
```

```python
# 11. Find the average daily revenue by dividing total_revenue by 7. Call this value average_daily_revenue and print it.
average_daily_revenue = total_revenue / 7
print(average_daily_revenue)
```

```python
# 12. Carly thinks she can bring in more customers by advertising all of the haircuts she offers that are under 30 dollars.
# Use a list comprehension to create a list called cuts_under_30 that includes the entry hairstyles[i] for each i for which new_prices[i] is less than 30.
# You can use range(len(new_prices)) in your list comprehension to make i go from 0 to the last index of new_prices.
# 13. Print

cuts_under_30 = [hairstyles[i] for i in range(len(new_prices)) if new_prices[i] < 30]

print(cuts_under_30)
```
